## Data Dictionary

- CBSA: <br>The name of the core bases statistical area (metropolitan area) where the monitoring site is located.                                   
- CBSA Code: <br>The code to identify individual CBSA's                             
- Year: <br>The year the data was collected                                  
- Days with AQI: <br> Total number of days where AQI data was reported                        
- Good Days: <br> Number of days in the year where the AQI value was between 0-50 (good air quality)                             
- Moderate Days: <br> Number of days in the year where the AQI value was between 51-100                          
- Unhealthy for Sensitive Groups Days: <br> Number of days in the year where the AQI value was between 101-150  
- Unhealthy Days: <br> Number of days in the year where the AQI value was between 151-200                        
- Very Unhealthy Days: <br> Number of days in the year where the AQI value was between 200-300                  
- Hazardous Days: <br> Number of days in the year where the AQI value was over 300                                           
- Max AQI: <br> The absolute highest daily AQI value recorded by any monitor in the CBSA during the year                                
- 90th Percentile AQI: <br> The AQI value that 90% of the daily values fall below.                  
- Median AQI:<br> The median daily AQI value recorded in the CBSA for the year                             
- Days CO: <br> Number of days that CO was the driving factor for AQI increase                                
- Days NO2: <br> Number of days that NO2 was the driving factor for AQI increase                             
- Days Ozone: <br> Number of days that O3 was the driving factor for AQI increase                            
- Days PM2.5: <br> Number of days that PM2.5 was the driving factor for AQI increase                             
- Days PM10: <br> Number of days that PM10 was the driving factor for AQI increase

## Visualizations  and Analyisis

```python
corrHM =[ "max_aqi", "90th_percentile_aqi", "median_aqi","days_co", "days_no2", "days_ozone", "days_pm25", "days_pm10"]
corr = df[corrHM].corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation of EPA AQI and contaminant data from 2016-2025")
plt.show()
```

This correlation heat map shows that there is potential correlation between the Max, median and 90th percentile AQI but there is very little to no correlation between the actual polutants. 

```python
# Looking at the average aqi over the years 2016-2025
avgAQI = (
    df.groupby("year")[["median_aqi"]]
    .mean()
    .reset_index()
)

plt.figure(figsize=(10,6))

sns.lineplot(
    data=avgAQI,
    x="year",
    y="median_aqi",
    marker="o",
    linewidth=3
)

plt.title("Average Median AQI Over Time")
plt.xlabel("Year")
plt.ylabel("Median AQI")

plt.show()
```

The median AQI shows that the Average median over 10 years is between 40 and 44 Which is in the Upper bound of the good air quality. This chart shows that there are drastic peaks indicating potential outside factors that might be affecting the data. I theorise that some of these factor might be COVID in late 2019 - 2021. However in perspective the median AQI has only varied on average 3 points over 10 years. The limitations to this chart and data is that It is essentually averaging averages, without daily data it might be harder to find outliers and create a more realistic representation. 

```python
corrHM =[ "good_days", "moderate_days", "unhealthy_days","unhealthy_for_sensitive_groups_days","very_unhealthy_days","hazardous_days", "days_co", "days_no2", "days_ozone", "days_pm25", "days_pm10"]
corr = df[corrHM].corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation of Days of high Pullutant Exposure and Number of Days of Different AQI Quality ")
plt.show()
```

This correlation chart shows that there is moderate correlation between moderate day AQI and days with high levels of PM 2.5. There is also inverse correlation between days with high PM2.5 and days with high O3.

```python
# Looking at total days of pollutants per year
pollutants = (df.groupby("year")[["days_co","days_no2","days_ozone","days_pm25","days_pm10"]].sum().reset_index())

pollutants
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=pollutants,
    x="year",
    y="days_co",
    marker="o",
    label="CO"
)
sns.lineplot(
    data=pollutants,
    x="year",
    y="days_no2",
    marker="o",
    label="NO2"
)
sns.lineplot(
    data=pollutants,
    x="year",
    y="days_ozone",
    marker="o",
    label="Ozone"
)
sns.lineplot(
    data=pollutants,
    x="year",
    y="days_pm25",
    marker="o",
    label="PM2.5"
)
sns.lineplot(
    data=pollutants,
    x="year",
    y="days_pm10",
    marker="o",
    label="PM10"
)
plt.title("EPA Pollutant Days over Time")
plt.xlabel("Year")
plt.ylabel("Number of Days from all CBSA's")
plt.xticks(pollutants["year"])
plt.legend(title="Pollutant")
plt.grid(True, alpha=0.3)
plt.show()
```

This graph shows that the two pullutants that cause the most change is AQI quality per year across all monitoring sites are Ozone or O3 and PM2.5 which is microscoptic air born particles. The two also appear to have an inverse relationship up until 2024. NO2 and CO stay consistant untill about 2022 and then they both seem to decrease. In 2024 PM10 Ozone and PM2.5 all begin the decrease.
